## 9. Order of operations (Operator Precedence)

```python
#1. Evaluate exponentiation first:
print(2 + 3 ** 2)       # = 2 + 9 = 11
print(5 * 2 ** 3)       # = 5 * 8 = 40
print(10 - 2 ** 2)      # = 10 - 4 = 6
```

```python
#2. Multiply and divide before adding and subtracting:
print(2 + 3 * 4)        # = 2 + 12 = 14
print(20 - 15 / 3)      # = 20 - 5.0 = 15.0
print(5 + 8 // 2)       # = 5 + 4 = 9
```

```python
#3. Advanced example:
print(2 * 3 + 4 * 5)    # = 6 + 20 = 26
print(100 / 4 - 3 * 2)  # = 25.0 - 6 = 19.0
print(2 ** 3 + 10 % 3)  # = 8 + 1 = 9
```

```python
#4. Left to right when precedence is equal (except exponentiation):
print(20 / 4 * 2)       # = 5.0 * 2 = 10.0
print(100 - 50 + 25)    # = 50 + 25 = 75
print(2 ** 3 ** 2)      # = 2 ** 9 = 512
```

```python
#Comparison example:
print("Without parentheses:", 2 + 3 * 4)
print("Following precedence:", 2 + (3 * 4))
```

## 20. String methods: upper() and lower()

```python
#upper() converts to uppercase:
print("hello".upper())              # HELLO
print("Python".upper())             # PYTHON
print("good morning".upper())       # GOOD MORNING
print("123abc".upper())             # 123ABC
```

```python
#lower() converts to lowercase:
print("HELLO".lower())              # hello
print("Python".lower())             # python
print("GOOD MORNING".lower())       # good morning
print("ABC123".lower())             # abc123
```

```python
#Usage example:
print("Name:", "john".upper())                    # JOHN
print("Last name:", "SMITH".lower())                # smith
print("Email:", "User@Email.COM".lower())         # user@email.com
print("Password:", "secret123".upper())           # SECRET123
```

```python
#Comparisons:
print("'Hello' == 'hello':", "Hello" == "hello")           # False
print("'Hello'.lower() == 'hello':", "Hello".lower() == "hello")   # True
print("'HELLO'.upper() == 'hello'.upper():", "HELLO".upper() == "hello".upper())   # True
```
