## 2. f-string

```python
#📝Practical example:
name = "Alice"
age = 25
height = 165.5

# Use an f-string
print(f"Name: {name}")
print(f"Age: {age} years")
print(f"Height: {height} cm")

# Longer sentence
print(f"Hello, {name} age {age} years and your height is {height} cm")
```

```python
#💰Arithmetic inside an f-string:
price = 100
quantity = 3
print(f"Total price: {price * quantity} THB")
```

```python
#⚖️Comparing f-strings with print arguments

name = "Alice"
age = 25
score = 95.5

#🔴 Traditional print arguments (more verbose):
print("Name:", name, "Age:", age, "years")
print("student", name, "scored", score, "points")

#🟢 F-string format (more readable):
print(f"Name: {name} Age: {age} years")
print(f"Student {name} scored {score} points")
```

```python
#📊 Practical example:
product = "iPhone"
price = 35000
discount = 10

# Traditional version
print("Product:", product, "Price:", price, "Discount:", discount, "%")

# F-string version
print(f"Product: {product} Price: {price} THB Discount: {discount}%")
print(f"Price after discount: {price - (price * discount / 100)} THB")
```

## 5. List Methods

```python
#append() example:
fruits = ["apple", "banana", "orange", "grape"]
fruits.append("mango")
print(fruits)  # ["apple", "banana", "orange", "grape", "mango"]
```

```python
#extend() example:
numbers = [10, 20, 30]
numbers.extend([40, 50])  # Adds multiple items
print(numbers)  # [10, 20, 30, 40, 50]
```

```python
# Compare append() and extend()
list1 = [10, 20, 30]
list1.append([40, 50])    # Adds the entire list as one element
print(list1)  # [10, 20, 30, [40, 50]]

list2 = [10, 20, 30]
list2.extend([40, 50])    # Adds each element separately
print(list2)  # [10, 20, 30, 40, 50]
```

```python
#insert() example:
colors = [10, 20, 30]
colors.insert(2, 40)  # Insert at index 2
print(colors)  # [10, 20, 40, 30]
```

```python
#remove() example:
fruits = [10, 20, 30, 40]
fruits.remove(10)  # Remove the value 10
print(fruits)  # [20, 30, 40]
```

```python
#pop() example:
numbers = [10, 20, 30, 40]
removed = numbers.pop(2)    # Remove the item at index 2
print(numbers)  # [10, 20, 40]
print(f"Removed: {removed}")  # Removed: 30
```

```python
# pop() without an index = remove the final item
last = numbers.pop()
print(numbers)  # [10, 20]
print(f"Last removed: {last}")  # Last removed: 40
```

```python
#clear() example:
temp = [10, 20, 30]
temp.clear()
print(temp)  # []
```

```python
#🛒 Shopping-list example:
shopping = []
shopping.append("milk")
shopping.append("bread")
shopping.insert(1, "eggs")
print(f"Shopping list: {shopping}")
```

```python
#count() example:
numbers = [10, 20, 30, 20]
count_20 = numbers.count(20)
print(f"There are {count_20} occurrences")  # There are 2 occurrences
```

```python
#index() example:
fruits = [10, 20, 30, 40]
position = fruits.index(40)
print(f"40 is at index {position}")  # 40 is at index 3
```

```python
#sort() example:
numbers = [30, 10, 40, 20]
numbers.sort()  # Sort in ascending order
print(numbers)  # [10, 20, 30, 40]

# Sort in descending order
numbers.sort(reverse=True)
print(numbers)  # [40, 30, 20, 10]
```

```python
#reverse() example:
data = [30, 20, 10]
data.reverse()  # Reverse the order
print(data)  # [10, 20, 30]
```

```python
#copy() example:
original = [10, 20, 30]
copied = original.copy()  # Create a new list
print(f"Original: {original}")
print(f"Copy: {copied}")

# Changing the copy does not affect the original
copied.append(40)
print(f"Original: {original}")  # [10, 20, 30]
print(f"Copy: {copied}")     # [10, 20, 30, 40]
```

```python
#Putting the methods together:
my_list = [30, 10, 20, 10]

print(f"Starting list: {my_list}")
my_list.append(40)              # [30, 10, 20, 10, 40]
my_list.insert(1, 15)           # [30, 15, 10, 20, 10, 40]
print(f"There are {my_list.count(10)} occurrences")  # 2 occurrences
my_list.remove(10)              # [30, 15, 20, 10, 40]
removed = my_list.pop(2)        # [30, 15, 10, 40]
my_list.sort()                  # [10, 15, 30, 40]
print(f"Result: {my_list}")
```

```python
#Practical example:
tasks = ["work", "study", "exercise", "read"]
tasks.remove("study")    # Finished studying
tasks.pop()              # No time left for reading
print(f"Remaining tasks: {tasks}")
```

## 9. Sets - Set operations

```python
#🔍 Example intersection():
set1 = {5, 2, 8}
set2 = {5, 1}
result = set1.intersection(set2)
print(result)  # {5}

# Or use &
result = set1 & set2
print(result)  # {5}
```

```python
#➖ Example difference():
set1 = {5, 2, 8}
set2 = {5, 1}
result = set1.difference(set2)
print(result)  # {8, 2}

# Or use -
result = set1 - set2
print(result)  # {8, 2}
```

```python
#🔗 Example union():
set1 = {1, 2, 3}
set2 = {3, 1}
result = set1.union(set2)
print(result)  # {1, 2, 3}

# Or use |
result = set1 | set2
print(result)  # {1, 2, 3}
```

```python
#🔍 Example issubset():
set1 = {1, -2}
set2 = {-2, 1, 3}
result = set1.issubset(set2)
print(result)  # True (set1 is a subset of set2)
```

```python
#📊 Example issuperset():
set1 = {-2, 1, 3}
set2 = {1, -2}
result = set1.issuperset(set2)
print(result)  # True (set1 contains every element of set2)
```

```python
#🚫 Example isdisjoint():
set1 = {-2, 1, 3}
set2 = {4, 6}
result = set1.isdisjoint(set2)
print(result)  # True (no common elements)

# Example returning False
set3 = {1, 5}
result = set1.isdisjoint(set3)
print(result)  # False (they share the element 1)
```

```python
#📊 Putting the methods together:
set_a = {1, 2, 3}
set_b = {3, 4, 5}

print(f"A ∪ B: {set_a | set_b}")        # {1, 2, 3, 4, 5}
print(f"A ∩ B: {set_a & set_b}")        # {3}
print(f"A - B: {set_a - set_b}")        # {1, 2}
print(f"B - A: {set_b - set_a}")        # {4, 5}
print(f"A ⊆ B: {set_a.issubset(set_b)}") # False
```

```python
## Real-world example
students_math = {'Alice', 'Bob', 'Charlie'}
students_science = {'Bob', 'Diana', 'Eve'}

# Students taking both subjects
both_subjects = students_math & students_science
print(f"Taking both subjects: {both_subjects}")  # {'Bob'}

# All students
all_students = students_math | students_science
print(f"All students: {all_students}")
```
